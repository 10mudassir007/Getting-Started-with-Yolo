```python
import kagglehub
from ultralytics import YOLO
import os
# Download the dataset (images)
path = kagglehub.dataset_download("shaunthesheep/microsoft-catsvsdogs-dataset")
path = path + os.sep + os.listdir(path)[0]
print("Downloaded to:", path)
```

```python
imgs,labels = [],[]
#os.path.join(path,os.listdir(path)[0])
for label in os.listdir(path):
  label_path = os.path.join(path,label)
  for img in os.listdir(label_path):
    imgs.append(os.path.join(label_path,img))
    labels.append(label)
```

```python
from PIL import Image

def load_img(path):
  try:
    return Image.open(path)
  except Exception as e:
    print(e)

images = [load_img(path) for path in imgs]
```

```python
wid = sum([x.width for x in images if x != None]) / len(images)
hei = sum([x.height for x in images if x != None]) / len(images)
wid, hei
```

```python
import shutil

src = '/kaggle/input/microsoft-catsvsdogs-dataset/PetImages'
dst = '/kaggle/working/PetImages'

if not os.path.exists(dst):
    shutil.copytree(src, dst)
    print("Copied!")

os.listdir(dst)
```

```python
dst
```

```python
model = YOLO("yolo26n-cls.pt")

model.train(
    data=dst,
    epochs=3,
    imgsz=224,
    batch=32,
    name='binary_classifier',
    device=0
)
```

```python
def prediction(path):
  img = Image.open(path)
  r = model.predict(img)[0]
  print(r.names[r.probs.top1])          # class name
  print(f"{r.probs.top1conf:.2%}")
prediction("/content/ginger-maine-coon-kitten-running-on-lawn-in-royalty-free-image-1719608142.jpg")
```
